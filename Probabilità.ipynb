```python
import random
```

```python
X = [random.random() for _ in range(1000)]
```

```python
plt.plot(X, '.');
```

```python
def next_lcg(x):
  return (A * x + C) % MAX

def lcg(n):
    x = SEED
    X = []
    for _ in range(n):
        x = next_lcg(x)
        X.append(x)
    return X
```

```python
A = 1664525
C = 1013904223
MAX = 2**32

SEED = 42

GOOD = lcg(1000)
```

```python
plt.plot(GOOD, '.');
```
